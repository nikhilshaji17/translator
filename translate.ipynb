```python
import getpass
import os

if not os.environ.get("SAMBANOVA_API_KEY"):
  os.environ["SAMBANOVA_API_KEY"] = getpass.getpass("Enter API key for Sambanova: ")
```

```python
from langchain_sambanova import ChatSambaNovaCloud

llm = ChatSambaNovaCloud(
    model="Meta-Llama-3.1-8B-Instruct",
    max_tokens=1024,
    temperature=0.7,
    top_p=0.01,
)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

to_translate = input("Enter the statement you want to translate:")

messages = [
    SystemMessage("Accept the HumanMessage, and convert it to French"),
    HumanMessage(to_translate),
]

llm.invoke(messages)
```

```python
from langchain_core.prompts import ChatPromptTemplate

system_template = "Translate the following from English into {language}"

prompt_template = ChatPromptTemplate.from_messages(
    [("system", system_template), ("user", "{text}")]
)
```

```python
language = input("Enter the language to translate to: ")
to_translate = input("Enter the string to translate: ")
prompt = prompt_template.invoke({"language": language, "text": to_translate})
response = llm.invoke(prompt)
print(response.content)
```
